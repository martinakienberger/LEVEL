## Análisis GLM

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
import seaborn as sns
```

#### Importar datos

```python
df = pd.read_pickle('Data_final')
#df = pd.get_dummies(df,columns=['Idioma', 'Nivel', 'Cluster'])
```

```python
df2 = pd.read_pickle('Data_completo')
#df2 = pd.get_dummies(df2,columns=['Idioma', 'Nivel', 'Cluster'])
```

### Porcentaje de palabras determinadas correctamente

```python
modelo_glm = smf.glm(
    formula="Pct_det ~ Estrategias_Por_Palabra + Num + C(Idioma, Treatment('Alemán')) + C(Nivel, Treatment('Principiantes')) + L_Num + C(Cluster, Treatment('Combi'))",
    data=df,
).fit()

print(modelo_glm.summary())
```

Solo sale significativo el nivel (menos aciertos en niveles intermedios)

### Comprensión lectora (resultado del resumen)

```python
modelo_glm = smf.glm(
    formula="Resumen ~ Pct_det + Estrategias_Por_Palabra + Num + C(Idioma, Treatment('Alemán')) + C(Nivel, Treatment('Principiantes')) + L_Num + C(Cluster, Treatment('Combi'))",
    data=df,
).fit()

print(modelo_glm.summary())
```

Salen significativos:
- Idioma: estudiantes chinos tienen puntuación menor
- Porcentaje de palabras determinadas correctamente: Por cada 10 puntos porcentuales adicionales de determinación correcta, la evaluación del resumen aumenta 0.156 puntos (en una escala de 0 a 3).
- Cluster "DiccTrad": Los estudiantes que dependen exclusivamente del diccionario/traductor obtienen 0.34 puntos menos en la comprensión global que quienes emplean una combinación amplia de herramientas (Combi), aun cuando se controla la determinación correcta del vocabulario.

### Porcentaje de palabras retenidas (resultado del test de vocabulario)

```python
modelo_glm = smf.glm(
    formula="Pct_ret ~ Estrategias_Por_Palabra + Resumen + Num + C(Idioma, Treatment('Alemán')) + C(Nivel, Treatment('Principiantes')) + L_Num + C(Cluster, Treatment('Combi'))",
    data=df2,
).fit()

print(modelo_glm.summary())
```

Salen significativos: 
- Idioma: estudiantes chinos tienen puntuación menor en el test de vocabulario
- Nivel: estudiantes en niveles intermedios tienen puntuación menor
- Resumen: Por cada punto adicional en la evaluación del resumen (escala 0-3), la retención de vocabulario aumenta 8.37 puntos porcentuales.
