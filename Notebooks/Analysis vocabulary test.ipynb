## Análisis del test de vocabulario

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

#### Importar datos


```python
df_aleman = pd.read_excel("Aleman_Voc.xlsx")
```

```python
df_chino = pd.read_excel("Chino_Voc.xlsx")
```

```python
df_ruso = pd.read_excel("Ruso_Voc.xlsx")
```

```python
df_aleman['Idioma'] = 'Alemán'
df_chino['Idioma'] = 'Chino'
df_ruso['Idioma'] = 'Ruso'
```

```python
datasets = [df_aleman,df_chino,df_ruso]
df = pd.concat(datasets, ignore_index=True)
df = df.fillna(0)
```

```python
df.head()
```

### Resultados del test de vocabulario

```python
columnas_totales_ev = [col for col in df.columns if col.endswith("_Ev")]

evaluaciones = []

for col in columnas_totales_ev:
    serie_limpia = df[col].dropna().astype(str).str.strip().str.replace("С", "C").str.replace('p','P').str.replace('´C','C')
    serie_valida = serie_limpia[
        ~serie_limpia.isin(["", "0", "nan", "None", "NaN"])
    ]

    evaluaciones.extend(serie_valida.tolist())

serie_evaluaciones = pd.Series(evaluaciones)
serie_evaluaciones.value_counts()
```

```python
serie_evaluaciones.value_counts(normalize=True).round(2)
```

```python
df['Ret_sum'] = df[columnas_totales_ev].replace({"C":1,
                                                "O":1,
                                                "P":0.5,
                                                "F":0,
                                                "N":0}).apply(pd.to_numeric, errors="coerce").sum(axis=1)
```

```python
df['Pct_ret'] = (df['Ret_sum']*10)/2
```
